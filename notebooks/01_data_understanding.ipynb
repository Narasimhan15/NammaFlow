Before deciding what to do with those two OneWay blocks, we need one final investigation:

**Q. Are the first OneWay records basically empty placeholder rows, or do they also contain genuine activity?**

```python
# ================================================================
# NAMMAFLOW — COMPARE THE TWO ONEWAY BLOCKS
# ================================================================

# Create a temporary copy — raw data remains untouched
ow = funnel_df[
    funnel_df["trip_category"] == "OneWay"
].copy()

# Identify first and second OneWay occurrence
# for each Ward × Vehicle combination
ow["oneway_occurrence"] = (
    ow.groupby(["ward_num", "vehicle_type"])
      .cumcount()
)

# Core operational metrics
metric_cols = [
    "srch_fr_e",
    "srch_which_got_e",
    "srch_fr_q",
    "srch_which_got_q",
    "booking",
    "ongng_ride",
    "done_ride",
    "rides",
    "cancel_ride",
    "drvr_cancel",
    "rider_cancel",
    "earning",
    "dist"
]

metric_cols = [c for c in metric_cols if c in ow.columns]


print("=" * 75)
print("1. ROW COUNT PER ONEWAY OCCURRENCE")
print("=" * 75)

print(
    ow["oneway_occurrence"]
    .value_counts()
    .sort_index()
)


# ------------------------------------------------
# Number of completely zero rows
# ------------------------------------------------

ow["all_core_metrics_zero"] = (
    ow[metric_cols].fillna(0).eq(0).all(axis=1)
)

print("\n" + "=" * 75)
print("2. ZERO vs NON-ZERO ROWS")
print("=" * 75)

print(
    pd.crosstab(
        ow["oneway_occurrence"],
        ow["all_core_metrics_zero"],
        margins=True
    )
)


# ------------------------------------------------
# Aggregate operational activity
# ------------------------------------------------

print("\n" + "=" * 75)
print("3. TOTAL ACTIVITY BY ONEWAY OCCURRENCE")
print("=" * 75)

activity_summary = (
    ow.groupby("oneway_occurrence")[metric_cols]
      .sum()
      .T
)

display(activity_summary)


# ------------------------------------------------
# Count exact equality between the pair
# ------------------------------------------------

print("\n" + "=" * 75)
print("4. EXACTLY IDENTICAL ONEWAY PAIRS")
print("=" * 75)

compare_cols = [
    c for c in funnel_df.columns
    if c not in ["ward_num", "trip_category", "vehicle_type"]
]

ow0 = (
    ow[ow["oneway_occurrence"] == 0]
    .set_index(["ward_num", "vehicle_type"])
    [compare_cols]
    .sort_index()
)

ow1 = (
    ow[ow["oneway_occurrence"] == 1]
    .set_index(["ward_num", "vehicle_type"])
    [compare_cols]
    .sort_index()
)

common_index = ow0.index.intersection(ow1.index)

same_pair = (
    ow0.loc[common_index]
    .reset_index(drop=True)
    .eq(
        ow1.loc[common_index]
        .reset_index(drop=True)
    )
    .all(axis=1)
)

print("Total OneWay pairs:", len(common_index))
print("Exactly identical pairs:", same_pair.sum())
print("Different pairs:", (~same_pair).sum())


# ------------------------------------------------
# Are there cases where BOTH blocks have activity?
# ------------------------------------------------

print("\n" + "=" * 75)
print("5. BOTH ONEWAY RECORDS HAVE ACTIVITY?")
print("=" * 75)

activity_flag = (
    ow.assign(
        has_activity=~ow["all_core_metrics_zero"]
    )
    .pivot_table(
        index=["ward_num", "vehicle_type"],
        columns="oneway_occurrence",
        values="has_activity",
        aggfunc="first"
    )
)

both_active = activity_flag[
    (activity_flag.get(0, False) == True) &
    (activity_flag.get(1, False) == True)
]

print(
    "Ward × vehicle pairs where BOTH OneWay records have activity:",
    len(both_active)
)

display(both_active.head(20))


# ------------------------------------------------
# Show examples where the two blocks differ
# ------------------------------------------------

print("\n" + "=" * 75)
print("6. SAMPLE DIFFERENT ONEWAY PAIRS")
print("=" * 75)

different_keys = common_index[~same_pair.values]

if len(different_keys) > 0:

    sample_keys = list(different_keys[:5])

    display(
        ow[
            ow.set_index(
                ["ward_num", "vehicle_type"]
            ).index.isin(sample_keys)
        ][
            ["ward_num",
             "vehicle_type",
             "oneway_occurrence"] + metric_cols
        ].sort_values(
            ["ward_num",
             "vehicle_type",
             "oneway_occurrence"]
        )
    )

else:
    print("No different pairs found.")
```

```python
# ================================================================
# NAMMAFLOW — CLEAN FUNNEL PLACEHOLDER RECORDS
# ================================================================

# Always preserve the original DataFrame
funnel_clean = funnel_df.copy()

# ------------------------------------------------
# Identify the two OneWay occurrences using
# ORIGINAL source row order
# ------------------------------------------------

funnel_clean["oneway_occurrence"] = -1

oneway_mask = (
    funnel_clean["trip_category"] == "OneWay"
)

funnel_clean.loc[
    oneway_mask,
    "oneway_occurrence"
] = (
    funnel_clean.loc[oneway_mask]
    .groupby(["ward_num", "vehicle_type"])
    .cumcount()
    .values
)


# ------------------------------------------------
# Remove occurrence 0:
# confirmed zero-activity placeholder rows
# ------------------------------------------------

placeholder_mask = (
    (funnel_clean["trip_category"] == "OneWay") &
    (funnel_clean["oneway_occurrence"] == 0)
)

print("=" * 75)
print("1. PLACEHOLDER ROWS TO REMOVE")
print("=" * 75)

print("Rows identified:", placeholder_mask.sum())


# Remove placeholders
funnel_clean = (
    funnel_clean.loc[~placeholder_mask]
    .drop(columns="oneway_occurrence")
    .reset_index(drop=True)
)


# ------------------------------------------------
# Validate cleaned dataset
# ------------------------------------------------

print("\n" + "=" * 75)
print("2. SHAPE AFTER CLEANING")
print("=" * 75)

print("Original shape:", funnel_df.shape)
print("Cleaned shape :", funnel_clean.shape)


print("\n" + "=" * 75)
print("3. TRIP CATEGORY COUNTS")
print("=" * 75)

print(
    funnel_clean["trip_category"]
    .value_counts()
)


print("\n" + "=" * 75)
print("4. ROWS PER WARD")
print("=" * 75)

print(
    funnel_clean.groupby("ward_num")
    .size()
    .value_counts()
    .sort_index()
)


print("\n" + "=" * 75)
print("5. CLEANED UNIQUE KEY CHECK")
print("=" * 75)

funnel_key = [
    "ward_num",
    "trip_category",
    "vehicle_type"
]

print(
    "Duplicate key rows:",
    funnel_clean.duplicated(
        subset=funnel_key
    ).sum()
)


print("\n" + "=" * 75)
print("6. EXACT DUPLICATE CHECK")
print("=" * 75)

print(
    "Exact duplicate rows:",
    funnel_clean.duplicated().sum()
)
```

The next thing we should tackle is the missing values. We should not simply fill them with 0, because many of the missing fields are rates, and a missing rate may simply mean its denominator was zero.

There is already one strong clue: in original city data, both variant columns had exactly 539 missing values, and earlier we saw exactly 539 rows where trip_category = All. We need to verify whether those fields are simply not applicable to the All category.

```python
#This does not change or impute anything. It only tells us why the values are missing.

# ================================================================
# NAMMAFLOW — MISSING VALUE DIAGNOSTICS
# ================================================================

# Preserve raw city data
city_clean = city_df.copy()


def missing_report(df, name):
    print("\n" + "=" * 75)
    print(f"{name} — MISSING VALUE REPORT")
    print("=" * 75)

    report = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2)
    })

    report = report[
        report["missing_count"] > 0
    ].sort_values(
        "missing_count",
        ascending=False
    )

    display(report)


# ------------------------------------------------
# 1. Overall missing-value reports
# ------------------------------------------------

missing_report(city_clean, "CITY")
missing_report(funnel_clean, "FUNNEL CLEAN")
missing_report(driver_df, "DRIVER")


# ------------------------------------------------
# 2. Investigate CITY variant-column missingness
# ------------------------------------------------

print("\n" + "=" * 75)
print("CITY — VARIANT COLUMN MISSINGNESS BY TRIP CATEGORY")
print("=" * 75)

variant_cols = [
    "srch_fr_q_variant",
    "srch_which_got_q_variant"
]

for col in variant_cols:

    if col in city_clean.columns:

        print(f"\n--- {col} ---")

        print(
            pd.crosstab(
                city_clean["trip_category"],
                city_clean[col].isna(),
                margins=True
            )
        )


# ------------------------------------------------
# 3. RATE COLUMN ZERO-DENOMINATOR DIAGNOSTIC
# ------------------------------------------------

def rate_null_check(df, rate_col, denominator_candidates):

    if rate_col not in df.columns:
        return

    print("\n" + "-" * 75)
    print(f"RATE: {rate_col}")
    print("-" * 75)

    print("Total missing:", df[rate_col].isna().sum())

    for denom in denominator_candidates:

        if denom not in df.columns:
            continue

        missing = df[rate_col].isna()
        zero = df[denom].fillna(0).eq(0)

        print(f"\nCandidate denominator: {denom}")

        print(
            "Missing rate + denominator = 0 :",
            (missing & zero).sum()
        )

        print(
            "Missing rate + denominator > 0 :",
            (missing & ~zero).sum()
        )

        print(
            "Rate present + denominator = 0 :",
            (~missing & zero).sum()
        )


print("\n" + "=" * 75)
print("CITY — RATE NULL PATTERNS")
print("=" * 75)

rate_null_check(
    city_clean,
    "cnvr_rate",
    ["srch_rqst", "booking", "rides"]
)

rate_null_check(
    city_clean,
    "q_accept_rate",
    ["srch_fr_q", "srch_which_got_q"]
)

rate_null_check(
    city_clean,
    "bkng_cancel_rate",
    ["booking", "cancel_ride"]
)


print("\n" + "=" * 75)
print("FUNNEL CLEAN — RATE NULL PATTERNS")
print("=" * 75)

rate_null_check(
    funnel_clean,
    "srch_to_e_rate",
    ["srch_fr_e", "srch_which_got_e"]
)

rate_null_check(
    funnel_clean,
    "e_to_q_srch_rate",
    ["srch_which_got_e", "srch_fr_q"]
)

rate_null_check(
    funnel_clean,
    "q_accept_rate",
    ["srch_fr_q", "srch_which_got_q"]
)

rate_null_check(
    funnel_clean,
    "q_to_bkng_rate",
    ["srch_which_got_q", "booking"]
)

rate_null_check(
    funnel_clean,
    "cnvr_rate",
    ["srch_fr_e", "booking", "rides"]
)

rate_null_check(
    funnel_clean,
    "bkng_cancel_rate",
    ["booking", "cancel_ride"]
)
```

The high rate-column missingness looks alarming at first, but it may actually be completely legitimate if the denominator is zero. That's what we'll check now.

```python
#This version avoids pd.crosstab(... margins=True) entirely, so the "All" category cannot conflict.

# ================================================================
# NAMMAFLOW — MISSING VALUE DIAGNOSTICS (CONTINUED / FIXED)
# ================================================================

# ------------------------------------------------
# 1. CITY variant-column missingness
# ------------------------------------------------

print("=" * 75)
print("1. CITY — VARIANT COLUMN MISSINGNESS BY TRIP CATEGORY")
print("=" * 75)

variant_cols = [
    "srch_fr_q_variant",
    "srch_which_got_q_variant"
]

for col in variant_cols:

    if col in city_clean.columns:

        temp = (
            city_clean
            .assign(is_missing=city_clean[col].isna())
            .groupby("trip_category")["is_missing"]
            .agg(
                total_rows="size",
                missing_rows="sum"
            )
        )

        temp["present_rows"] = (
            temp["total_rows"] -
            temp["missing_rows"]
        )

        temp["missing_pct"] = (
            temp["missing_rows"] /
            temp["total_rows"] * 100
        ).round(2)

        print(f"\n--- {col} ---")
        display(temp)


# ------------------------------------------------
# Helper function for rate diagnostics
# ------------------------------------------------

def rate_null_check(df, rate_col, denominator_candidates):

    if rate_col not in df.columns:
        return

    print("\n" + "-" * 75)
    print(f"RATE: {rate_col}")
    print("-" * 75)

    missing = df[rate_col].isna()

    print("Total rows   :", len(df))
    print("Missing rate :", missing.sum())
    print("Present rate :", (~missing).sum())

    for denom in denominator_candidates:

        if denom not in df.columns:
            continue

        zero = df[denom].fillna(0).eq(0)

        print(f"\nCandidate denominator → {denom}")

        print(
            "Missing rate & denominator = 0 :",
            (missing & zero).sum()
        )

        print(
            "Missing rate & denominator > 0 :",
            (missing & ~zero).sum()
        )

        print(
            "Rate present & denominator = 0 :",
            (~missing & zero).sum()
        )


# ------------------------------------------------
# 2. CITY rate diagnostics
# ------------------------------------------------

print("\n" + "=" * 75)
print("2. CITY — RATE NULL PATTERNS")
print("=" * 75)

rate_null_check(
    city_clean,
    "cnvr_rate",
    ["srch_rqst", "booking", "rides"]
)

rate_null_check(
    city_clean,
    "q_accept_rate",
    ["srch_fr_q", "srch_which_got_q"]
)

rate_null_check(
    city_clean,
    "bkng_cancel_rate",
    ["booking", "cancel_ride"]
)


# ------------------------------------------------
# 3. FUNNEL rate diagnostics
# ------------------------------------------------

print("\n" + "=" * 75)
print("3. FUNNEL CLEAN — RATE NULL PATTERNS")
print("=" * 75)

rate_null_check(
    funnel_clean,
    "srch_to_e_rate",
    ["srch_fr_e", "srch_which_got_e"]
)

rate_null_check(
    funnel_clean,
    "e_to_q_srch_rate",
    ["srch_which_got_e", "srch_fr_q"]
)

rate_null_check(
    funnel_clean,
    "q_accept_rate",
    ["srch_fr_q", "srch_which_got_q"]
)

rate_null_check(
    funnel_clean,
    "q_to_bkng_rate",
    ["srch_which_got_q", "booking"]
)

rate_null_check(
    funnel_clean,
    "cnvr_rate",
    ["srch_fr_e", "booking", "rides"]
)

rate_null_check(
    funnel_clean,
    "bkng_cancel_rate",
    ["booking", "cancel_ride"]
)
```

The missing values are not random data-quality problems. They are mostly mathematically undefined rates.

For **city_df**, the pattern is exact:
- cnvr_rate is missing 4,150 times, and all 4,150 occur where srch_rqst = 0.

- q_accept_rate is missing 4,353 times, and all occur where srch_fr_q = 0.

- bkng_cancel_rate is missing 4,347 times, and all occur where booking = 0.

- Both *_variant columns are missing exactly 539 times, and every one of those rows has trip_category = "All".

For **funnel_clean** it is equally clean:
srch_to_e_rate   missing when srch_fr_e = 0
e_to_q_srch_rate missing when srch_which_got_e = 0
q_accept_rate    missing when srch_fr_q = 0
q_to_bkng_rate   missing when srch_which_got_q = 0
cnvr_rate        missing when srch_fr_e = 0
bkng_cancel_rate missing when booking = 0

So we should NOT use mean, median, KNN, or zero-fill these rate columns. A rate like booking cancellations / bookings simply has no mathematical value when bookings are zero.

--> Keep structurally undefined rates as NaN. Treat the two variant fields as “not applicable” for trip_category = All.

The next crucial step:

**Q. what does vehicle_type = "All" mean?**

```python
#calculating hourly booking totals
city_bengaluru.groupby("hour")["booking"].sum()
```

```python
# ================================================================
# NAMMAFLOW — VERIFY WHETHER vehicle_type = "All" IS AGGREGATED
# ================================================================

metrics = [
    "srch_rqst",
    "booking",
    "rides",
    "earning"
]

group_cols = [
    "date",
    "hour",
    "trip_category",
    "booking_type"
]

# All-vehicle rows
all_rows = (
    city_bengaluru[
        city_bengaluru["vehicle_type"] == "All"
    ][group_cols + metrics]
    .copy()
)

# Individual vehicle rows
detail_rows = (
    city_bengaluru[
        city_bengaluru["vehicle_type"] != "All"
    ]
    .copy()
)

# Sum Auto + Cab etc. at the same grain
detail_sum = (
    detail_rows
    .groupby(group_cols, as_index=False)[metrics]
    .sum()
)

# Compare
compare = all_rows.merge(
    detail_sum,
    on=group_cols,
    how="left",
    suffixes=("_all", "_detail")
)

print("=" * 70)
print("ALL VEHICLE vs SUM OF DETAIL VEHICLES")
print("=" * 70)

for metric in metrics:

    valid = compare[f"{metric}_detail"].notna()

    matches = (
        compare.loc[valid, f"{metric}_all"]
        ==
        compare.loc[valid, f"{metric}_detail"]
    )

    print(f"\n{metric}")
    print("Rows compared :", valid.sum())
    print("Matches       :", matches.sum())
    print("Different     :", (~matches).sum())

    if valid.sum() > 0:
        print(
            "Match %       :",
            round(matches.mean() * 100, 2)
        )


print("\n" + "=" * 70)
print("SAMPLE COMPARISON")
print("=" * 70)

display(
    compare[
        group_cols +
        [
            "booking_all",
            "booking_detail",
            "rides_all",
            "rides_detail",
            "earning_all",
            "earning_detail"
        ]
    ].head(20)
)
```

So we must not treat All as Auto + Cab.

What this likely means structurally is that the All rows are being used for a different stage of the funnel — especially the search/demand side — while vehicle-specific rows carry much of the booking, ride, cancellation and earnings activity.

That would also explain why srch_rqst had 0% exact matches:
overall search requests do not necessarily equal the sum of vehicle-specific search/request figures.

So the correct rule is not “remove All” and also not “sum everything together.”

We need one final semantic profile of All vs Auto/Cab. This will tell us which metrics should come from which row type.

```python
# ================================================================
# NAMMAFLOW — PROFILE 'ALL' VS DETAIL VEHICLE ROWS
# ================================================================

profile_metrics = [
    "srch_rqst",
    "srch_which_got_e",
    "srch_fr_q",
    "srch_which_got_q",
    "booking",
    "rides",
    "done_ride",
    "earning",
    "cancel_ride",
    "drvr_cancel",
    "rider_cancel"
]

profile_metrics = [
    c for c in profile_metrics
    if c in city_bengaluru.columns
]


print("=" * 75)
print("1. ACTIVITY PROFILE — ALL VEHICLE ROWS")
print("=" * 75)

all_profile = []

all_df = city_bengaluru[
    city_bengaluru["vehicle_type"] == "All"
].copy()

for col in profile_metrics:

    all_profile.append({
        "metric": col,
        "rows": len(all_df),
        "non_zero_rows": all_df[col].fillna(0).ne(0).sum(),
        "zero_rows": all_df[col].fillna(0).eq(0).sum(),
        "total": all_df[col].fillna(0).sum()
    })

display(pd.DataFrame(all_profile))


print("\n" + "=" * 75)
print("2. ACTIVITY PROFILE — DETAIL VEHICLE ROWS")
print("=" * 75)

detail_df = city_bengaluru[
    city_bengaluru["vehicle_type"] != "All"
].copy()

detail_profile = []

for col in profile_metrics:

    detail_profile.append({
        "metric": col,
        "rows": len(detail_df),
        "non_zero_rows": detail_df[col].fillna(0).ne(0).sum(),
        "zero_rows": detail_df[col].fillna(0).eq(0).sum(),
        "total": detail_df[col].fillna(0).sum()
    })

display(pd.DataFrame(detail_profile))


print("\n" + "=" * 75)
print("3. SAMPLE ALL ROWS WITH FULL FUNNEL METRICS")
print("=" * 75)

display(
    all_df[
        [
            "date",
            "hour",
            "trip_category",
            "vehicle_type",
            "booking_type"
        ] + profile_metrics
    ].head(20)
)


print("\n" + "=" * 75)
print("4. SAMPLE DETAIL ROWS WITH FULL FUNNEL METRICS")
print("=" * 75)

display(
    detail_df[
        [
            "date",
            "hour",
            "trip_category",
            "vehicle_type",
            "booking_type"
        ] + profile_metrics
    ].head(20)
)
```

For NammaFlow, our rule from now onward is:
**Demand / funnel analysis: use vehicle_type == "All".**

**Bookings, completed rides, cancellations, earnings and vehicle analysis: use vehicle_type != "All".**

And we should never blindly sum "All" + Auto + Cab.

That also means the hourly bookings I calculated earlier by grouping the full city_bengaluru table should be ignored. We'll rebuild hourly KPIs correctly from the operational rows.

```python
# ================================================================
# NAMMAFLOW — FINAL ANALYTICAL DATASETS
# ================================================================

# ------------------------------------------------
# 1. DEMAND / SEARCH DATA
# ------------------------------------------------

city_demand = (
    city_bengaluru[
        city_bengaluru["vehicle_type"] == "All"
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------
# 2. VEHICLE / OPERATIONAL DATA
# ------------------------------------------------

city_operations = (
    city_bengaluru[
        city_bengaluru["vehicle_type"] != "All"
    ]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------
# 3. Mark complete vs partial dates
# ------------------------------------------------

hours_per_date = (
    city_bengaluru
    .groupby("date")["hour"]
    .nunique()
)

complete_dates = (
    hours_per_date[
        hours_per_date == 24
    ]
    .index
)

city_demand["is_complete_day"] = (
    city_demand["date"].isin(complete_dates)
)

city_operations["is_complete_day"] = (
    city_operations["date"].isin(complete_dates)
)


# ------------------------------------------------
# 4. Validation
# ------------------------------------------------

print("=" * 75)
print("NAMMAFLOW — FINAL DATA UNDERSTANDING VALIDATION")
print("=" * 75)

print("\nCITY — RAW BENGALURU")
print("Shape:", city_bengaluru.shape)

print("\nCITY — DEMAND DATA")
print("Shape:", city_demand.shape)
print(
    "Vehicle types:",
    city_demand["vehicle_type"].unique()
)

print("\nCITY — OPERATIONS DATA")
print("Shape:", city_operations.shape)
print(
    "Vehicle types:",
    city_operations["vehicle_type"].unique()
)

print("\nWARD FUNNEL CLEAN")
print("Shape:", funnel_clean.shape)

print("\nDRIVER DATA")
print("Shape:", driver_df.shape)


print("\n" + "=" * 75)
print("COMPLETE / PARTIAL DATE STATUS")
print("=" * 75)

date_status = (
    city_bengaluru
    .groupby("date")["hour"]
    .nunique()
    .reset_index(name="hours_available")
)

date_status["day_status"] = (
    date_status["hours_available"]
    .apply(
        lambda x: "Complete"
        if x == 24
        else "Partial"
    )
)

display(date_status)


print("\n" + "=" * 75)
print("FINAL DUPLICATE VALIDATION")
print("=" * 75)

print(
    "Demand exact duplicates:",
    city_demand.duplicated().sum()
)

print(
    "Operations exact duplicates:",
    city_operations.duplicated().sum()
)

print(
    "Funnel exact duplicates:",
    funnel_clean.duplicated().sum()
)

print(
    "Driver exact duplicates:",
    driver_df.duplicated().sum()
)
```
